# Compound Interest: 
### Write a Python program to calculate compound interest. Use the formula: 
### A = P * (1 + r/n)^(n-t)
Where: 
### A= total amount 
### P = principal amount 
### r = annual interest rate (decimal) 
### t = time in years 
### n = number of times interest is compounded per year
### Take P, r, n, and t as inputs from the user.

```python
P = float(input("write principle amount :"))
t = float(input("write time in years :"))
r = float(input("write annual interest in decimal :"))
n = float(input("write number of times interest is compounded per year :"))
A = P*((1+r/n))**(n-t)
print (A)
```

# Work Done by a Force:
### rite a Python program to calculate the work done by a force. Use the formula:
### where:
### W = work done
### f= force
### d = distance
### theta = angle between force and direction of movement (in degrees)
### Take f, d, and θ as inputs from the user.


```python
import math
f = float(input("Enter the force (in Newtons): "))
d = float(input("Enter the distance (in meters): "))
theta = float(input("Enter the angle theta (in degrees): "))
theta_radians = math.radians(theta)
W = f * d * math.cos(theta_radians)
print (W)
```

# Heat Transfer:
### Write a Python program to calculate the amount of heat transferred. Use the formula:
### 𝑄 = 𝑚𝑐 (∆T)
### where:
### Q= heat transfer
### m = mass
### c = specific heat capacity
### ΔT
### Take m, c, T as inputs from the user

```python
m = float(input("Enter the mass (m in kg): "))
c = float(input("Enter the specific heat capacity (c): "))
dT = float(input("Enter the temperature change (Delta T in Celsius): "))
Q = m * c * dT
print(Q)
```
