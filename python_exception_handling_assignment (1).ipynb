## Create a class-based bank account system. Handle deposits and withdrawals using custom exceptions for invalid transaction amounts and insufficient balance.

```python
class BankAccount:
    def __init__(self, balance=0):
        self.balance = balance
    def deposit(self, amount):
        if amount <= 0:
            raise InvalidAmountError("Deposit amount must be greater than zero.")
        self.balance += amount
        print(f"Deposited Rs. {amount} | New Balance: Rs. {self.balance}")
    def withdraw(self, amount):
        if amount <= 0:
            raise InvalidAmountError("Withdrawal amount must be greater than zero.")
        if amount > self.balance:
            raise InsufficientBalanceError("Insufficient balance in your account.")
        self.balance -= amount
        print(f"Withdrew Rs. {amount} | Remaining Balance: Rs. {self.balance}")
account = BankAccount(5000)
try:
    account.deposit(2000)
    account.withdraw(3000)
    account.withdraw(10000)
except InvalidAmountError as e:
    print("Amount Error:", e)
except InsufficientBalanceError as e:
    print("Balance Error:", e)
```

## Create an e-commerce order system that handles invalid products, invalid quantities, insufficient stock, and invalid payments. Use custom exceptions for business-rule failures.

```python
class InvalidProductError(Exception):
    pass
class InvalidQuantityError(Exception):
    pass
class InsufficientStockError(Exception):
    pass
class InvalidPaymentError(Exception):
    pass
products = {
    "laptop": {"price": 80000, "stock": 5},
    "mouse": {"price": 1500, "stock": 10},
    "keyboard": {"price": 3000, "stock": 3}
}
try:
    item = input("Enter product name (laptop / mouse / keyboard): ").strip().lower()
    if item not in products:
        raise InvalidProductError("Product does not exist.")
    qty = int(input("Enter quantity: "))
    if qty <= 0:
        raise InvalidQuantityError("Quantity must be greater than zero.")
    if qty > products[item]["stock"]:
        raise InsufficientStockError("Not enough stock available.")
    total_bill = products[item]["price"] * qty
    print(f"Total Bill: Rs. {total_bill}")
    payment = float(input("Enter payment amount: "))
    if payment < total_bill:
        raise InvalidPaymentError("Payment amount is less than total bill.")
    products[item]["stock"] -= qty
    change = payment - total_bill
    print(f"Order successful! Change returned: Rs. {change}")
except InvalidProductError as e:
    print("Product Error:", e)
except InvalidQuantityError as e:
    print("Quantity Error:", e)
except InsufficientStockError as e:
    print("Stock Error:", e)
except InvalidPaymentError as e:
    print("Payment Error:", e)
except ValueError:
    print("Input Error: Please enter valid numeric values.")
```

## ### University Course Registration System

### Goal:
Create a program that allows students to register for university courses.

### Steps:
- Store students and courses.
- Check whether the student exists.
- Check whether the course exists.
- Prevent duplicate registration.
- Check course capacity.
- Check prerequisites.
- Create and handle suitable custom exceptions.

### Example:
Student ID: 102
Course: Python
→ Registration successful

```python
class StudentNotFoundError(Exception):
    pass
class CourseNotFoundError(Exception):
    pass
class DuplicateRegistrationError(Exception):
    pass
class CourseFullError(Exception):
    pass
class PrerequisiteError(Exception):
    pass
students = {
    101: {"name": "Ali", "completed": ["Basic Math"], "registered": []},
    102: {"name": "Sara", "completed": ["Basic Math", "Programming Fundamentals"], "registered": []}
}
courses = {
    "Python": {"capacity": 2, "enrolled": 1, "prereq": "Basic Math"},
    "Data Science": {"capacity": 1, "enrolled": 1, "prereq": "Python"}
}
try :
    sid = int(input("Enter the student id :"))
    if sid not in students:
        raise StudentNotFoundError ("Incorrect id, student not found")
    course = input("Enter the course name : ")
    if course not in courses:
        raise CourseNotFoundError ("Course are not available")
    if course in students[sid]["registered"]:
        raise DuplicateRegistrationError("Student is already registered in this course.")
    if courses[course]["enrolled"] >=courses[course]["capacity"]:
        raise CourseFullError ("No seats are available")
    req = courses[course]["prereq"]
    if req not in students[sid]["completed"]:
        raise PrerequisiteError ("Complete the Prerequisite course first")
    students[sid]["registered"].append(course)
    courses[course]["enrolled"] +=1
    print(f"Student ID: {sid} Course: {course} -> Registration successful")
except StudentNotFoundError as e:
    print("Student Error:", e)
except CourseNotFoundError as e:
    print("Course Error:", e)
except DuplicateRegistrationError as e:
    print("Duplicate Error:", e)
except CourseFullError as e:
    print("Capacity Error:", e)
except PrerequisiteError as e:
    print("Prerequisite Error:", e)
except ValueError:
    print("Input Error: Student ID must be a number.")
```

## ### Hospital Appointment System

### Goal:
Create a program that manages hospital appointments.

### Steps:
- Store patients and doctors.
- Validate patient and doctor information.
- Check doctor availability.
- Prevent duplicate appointment times.
- Create custom exceptions for important business errors.
- Handle invalid input without crashing.

### Example:
Patient ID: P102
Doctor: D05
Time: 10:30 AM
→ Appointment booked successfully

```python
class PatientNotFoundError(Exception):
    pass
class DoctorNotFoundError(Exception):
    pass
class DoctorUnavailableError(Exception):
    pass
class DuplicateAppointmentError(Exception):
    pass
patients = {
    "P101": "Ali",
    "P102": "Sara"
}
doctors = {
    "D05": {"name": "Dr. Ahmed", "available_slots": ["10:00 AM", "10:30 AM", "11:00 AM"], "booked": ["10:00 AM"]},
    "D06": {"name": "Dr. Ayesha", "available_slots": ["02:00 PM", "03:00 PM"], "booked": []}
}
try:
    pid = input("Enter Patient ID (P101 / P102): ").strip().upper()
    if not pid:
        raise ValueError("Patient ID cannot be empty.")
    if pid not in patients:
        raise PatientNotFoundError("Patient ID does not exist.")
    did = input("Enter Doctor ID (D05 / D06): ").strip().upper()
    if not did:
        raise ValueError("Doctor ID cannot be empty.")
    if did not in doctors:
        raise DoctorNotFoundError("Doctor ID does not exist.")
    time = input("Enter Appointment Time (e.g., 10:30 AM): ").strip().upper()
    if not time:
        raise ValueError("Appointment time cannot be empty.")
    if time not in doctors[did]["available_slots"]:
        raise DoctorUnavailableError("Doctor is not available at this time.")
    if time in doctors[did]["booked"]:
        raise DuplicateAppointmentError("This time slot is already booked.")
    doctors[did]["booked"].append(time)
    print(f"Patient ID: {pid} Doctor: {did} Time: {time} -> Appointment booked successfully")
except PatientNotFoundError as e:
    print("Patient Error:", e)
except DoctorNotFoundError as e:
    print("Doctor Error:", e)
except DoctorUnavailableError as e:
    print("Availability Error:", e)
except DuplicateAppointmentError as e:
    print("Duplicate Error:", e)
except ValueError as e:
    print("Input Error:", e)
```

## ### Digital Wallet System

### Goal:
Create a class-based digital wallet application.

### Steps:
- Validate wallet ID and PIN.
- Handle inactive accounts.
- Validate transfer amounts.
- Check balance.
- Enforce a transfer limit.
- Create multiple custom exceptions.
- Make sure failed transfers do not change wallet balances.

### Example:
Transfer: Rs. 2,000
→ Transfer successful

```python
class WalletNotFoundError(Exception):
    pass
class InvalidPinError(Exception):
    pass
class InactiveAccountError(Exception):
    pass
class InvalidAmountError(Exception):
    pass
class TransferLimitError(Exception):
    pass
class InsufficientBalanceError(Exception):
    pass
class DigitalWallet:
    def __init__(self, accounts, limit=10000):
        self.accounts = accounts
        self.limit = limit
    def transfer(self, sender_id, pin, receiver_id, amount):
        if sender_id not in self.accounts or receiver_id not in self.accounts:
            raise WalletNotFoundError("Sender or Receiver wallet ID not found.")
        if sender_id == receiver_id:
            raise InvalidAmountError("Cannot transfer money to the same wallet.")
        sender = self.accounts[sender_id]
        receiver = self.accounts[receiver_id]
        if sender["pin"] != pin:
            raise InvalidPinError("Incorrect PIN entered.")
        if not sender["active"] or not receiver["active"]:
            raise InactiveAccountError("Sender or Receiver account is inactive.")
        if amount <= 0:
            raise InvalidAmountError("Transfer amount must be greater than zero.")
        if amount > self.limit:
            raise TransferLimitError(f"Transfer exceeds limit of Rs. {self.limit}.")
        if amount > sender["balance"]:
            raise InsufficientBalanceError("Insufficient balance in wallet.")
        sender["balance"] -= amount
        receiver["balance"] += amount
        print(f"Transfer: Rs. {amount:,.0f} -> Transfer successful")
wallets = {
    "W101": {"pin": "1234", "balance": 15000, "active": True},
    "W102": {"pin": "5678", "balance": 3000, "active": True},
    "W103": {"pin": "0000", "balance": 5000, "active": False}
}
app = DigitalWallet(wallets)
try:
    sid = input("Enter Your Wallet ID (W101 / W102 / W103): ").strip().upper()
    pin = input("Enter PIN: ").strip()
    rid = input("Enter Receiver Wallet ID: ").strip().upper()
    amt = float(input("Enter Transfer Amount: "))
    app.transfer(sid, pin, rid, amt)
except WalletNotFoundError as e:
    print("Wallet Error:", e)
except InvalidPinError as e:
    print("PIN Error:", e)
except InactiveAccountError as e:
    print("Account Error:", e)
except InvalidAmountError as e:
    print("Amount Error:", e)
except TransferLimitError as e:
    print("Limit Error:", e)
except InsufficientBalanceError as e:
    print("Balance Error:", e)
except ValueError:
    print("Input Error: Please enter a valid numeric amount.")
```

## ### Retail Inventory and Sales System

### Task:
Create a menu-driven inventory and sales application.

The program should allow the user to:
- Add stock
- View inventory
- Sell products
- Search for products

Handle invalid input, missing products, invalid quantities, and insufficient stock. Use `raise`, custom exceptions, `else`, and `finally` where appropriate.

The application should continue running after recoverable errors.

```python
class ProductNotFoundError(Exception):
    pass
class InvalidQuantityError(Exception):
    pass
class InsufficientStockError(Exception):
    pass
inventory = {
    "apple": {"price": 200, "stock": 50},
    "milk": {"price": 180, "stock": 30},
    "bread": {"price": 120, "stock": 20}
}
while True:
    print("\n--- Retail Inventory & Sales Menu ---")
    print("1. Add Stock | 2. View Inventory | 3. Sell Product | 4. Search Product | 5. Exit")
    choice = input("Enter choice (1-5): ").strip()
    if choice == "5":
        print("Thanks for using app")
        break
    try :
        if choice == "1":
            name  = input("Enter product name: ")
            if not name or name.isdigit():
                raise ValueError ("Invalid product name")
            qty = int(input("Enter the qty :"))
            if not qty :
                raise ValueError ("Invalid Qty")
            if qty <= 0:
                raise InvalidQuantityError ("Quantity must be greater than zero")
            if name in inventory:
                inventory[name]["stock"] += qty
            else:
                price = float(input("New product! Enter unit price: "))
                if price <= 0:
                    raise ValueError("Price must be greater than zero.")
                inventory[name] = {"Price":price,"stock" : qty}
        elif choice == "2":
            for pro, info in inventory:
                print(f"Product {pro} : Price {info["price"]} : Stock {info["stock"]}")
        elif choice == "3":
            name  = input("Enter product name: ")
            if name not in inventory :
                raise ProductNotFoundError ("Product not in a store")
            qty = int(input("Enter the qty :"))
            if qty <= 0 :
                raise InvalidQuantityError ("Quantity must be greater than zero")
            if qty > inventory[name]["stock"]:
                raise InsufficientStockError ("Not have a enough stock")
            inventory[name]["stock"] -= qty
            total = inventory[name]["price"] * qty
            print(f"Sold {qty} x {name.capitalize()} | Total Bill: Rs. {total}")
        elif choice == "4" :
            name  = input("Enter product name: ")
            if name not in inventory:
                raise ProductNotFoundError ("Prodcut not found")
            info = inventory[name]
            print(f"Found: {name.capitalize()} | Price: Rs. {info['price']} | Stock: {info['stock']}")
        else:
            raise ValueError("Invalid menu choice. Please select 1 to 5.")
    except ProductNotFoundError as e :
        print(f"Error {e}")
    except InvalidQuantityError as e:
        print("Quantity Error:", e)
    except InsufficientStockError as e:
        print("Stock Error:", e)
    except ValueError as e:
        print("Input Error:", e)
    else:
        print("Operation completed successfully.")
    finally:
        print("Ready for next action...")
```
