## Write a program to detect suspicious banking transactions.
Take input:
* transaction_amount
* location_match (Yes/No)
* time_of_transaction(Day/Night)
* previous_transactions_count
* device_verified (Yes/No)
#### Rules:
* amount > 100,000 → suspicious
* location mismatch + high amount → fraud alert
* night transaction + unverified device → high risk
* many transactions (>10 in 1 hour) → freeze account
* otherwise → normal transaction

```python
amount = float(input("Enter transaction amount: "))
location_match = input("Is location matching?: ")
time_of_transaction = input("Transaction time: ")
prev_count = int(input("Enter transactions count in last 1 hour: "))
device_verified = input("Is device verified?: ")
if prev_count > 10:
    print("FREEZE ACCOUNT")
elif location_match == "No" and amount > 100000:
    print("FRAUD ALERT")
elif time_of_transaction == "Night" and device_verified == "No":
    print("HIGH RISK")
elif amount > 100000:
    print("SUSPICIOUS TRANSACTION")
else:
    print("NORMAL TRANSACTION")
```

## Write a program to classify a person’s risk level in a police database.
Take input:
* criminal_record (Yes/No)
* number_of_cases
* travel_history_flagged (Yes/No)
* financial_activity_score
* last_activity_days
#### Rules:
* criminal_record = Yes AND cases > 3 → High Risk
* flagged travel + suspicious financial score → High Risk
* no activity for long time (>365 days) → Monitor
* clean record + normal activity → Low Risk
* mixed signals → Medium Risk

```python
criminal_record = input("Criminal record?: ")
number_of_cases = int(input("Enter number of cases: "))
travel_history_flagged = input("Travel history flagged?: ")
financial_activity_score = int(input("Enter financial activity score: "))
last_activity_days = int(input("Enter days since last activity: "))
if criminal_record == "Yes" and number_of_cases > 3:
    print("High Risk")
elif travel_history_flagged == "Yes" and financial_activity_score > 75:
    print("High Risk")
elif last_activity_days > 365:
    print("Monitor")
elif criminal_record == "No" and number_of_cases == 0 and financial_activity_score <= 50:
    print("Low Risk")
else:
    print("Medium Risk")
```
