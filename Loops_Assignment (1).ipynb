## Create a menu-driven banking system using loops:
* Create Account
* Deposit
* Withdraw
* Check Balance
* Exit

```python
#### balance = 0
account_created = False
name = ""
for i in range(1000):
    print("--- BANKING SYSTEM MENU ---")
    print("1. Create Account")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Check Balance")
    print("5. Exit")

    choice = int(input("Enter the choice"))
    if choice == 1 :
        name = input("What is your name ")
        balance = int(input("Initial deposit amount : "))
        account_created = True
        print (f"Account created successfully for {name}!")
    elif choice == 2 :
        if account_created:
            amount = int(input("Enter deposit amount : "))
            amount += balance
            print (f"{amount} deposited.")
        else :
            print("Create your accout first")
    elif choice == 3 :
        if account_created:
            amount = int(input("Enter the withdrawal amount : "))
            amount -= balance
        else:
            print("Create your accout first")
    elif choice == 4 :
        if account_created:
            print(f"You account balance is {balance}")
        else:
            print("Create your accout first")
    elif choice == 5 :
        print("Thank you for using our banking system. Goodbye!")
        break
    else :
        print("Invalid choice! Please select between 1-5.")
```

## Hotel Room Booking Tracker
### Goal:

    Record bookings for rooms on each floor.

### Steps:

    Outer loop → for 3 floors.
    
    Inner loop → for 4 rooms per floor.
    
    Ask user if booked (Y/N).
    
    Count total booked and available rooms.

    Room 4: Y

```python
total_booked = 0
total_available = 0

for floor in range(1, 4):
    print(f"Floor {floor}:")
    floor_booked = 0
    
    for room in range(1, 5):
        status = input(f"Room {room}: ")
        if status == "Y" or status == "y":
            floor_booked += 1
            total_booked += 1
        else:
            total_available += 1
            
    print(f"→ Output: Floor {floor} = {floor_booked} booked")

print(f"\nOverall Summary: Total Booked = {total_booked}, Total Available = {total_available}")
```

## Invoice Tax Calculator

## Student Result Sheet
### Goal:

    Generate report cards for each student in a class.

### Steps:

    Outer loop → 5 students.
    
    Inner loop → 4 subjects.

    If marks < 33 → print “Failed in Subject”.

    Calculate total and grade per student.

### Example:
    Student 1: [80, 90, 70, 85] → Grade: A  
    Student 2: [50, 32, 45, 40] → Failed in Subject 2

```python
for student in range(1, 6):
    print(f"\nStudent {student}:")
    marks_list = []
    failed = False
    
    for subject in range(1, 5):
        marks = int(input(f"Subject {subject}: "))
        marks_list.append(marks)
        
        if marks < 33:
            print(f"→ Failed in Subject {subject}")
            failed = True
            
    if not failed:
        total = sum(marks_list)
        percentage = total / 4
        
        if percentage >= 80:
            grade = "A"
        elif percentage >= 60:
            grade = "B"
        elif percentage >= 45:
            grade = "C"
        else:
            grade = "D"
            
        print(f"→ {marks_list} → Grade: {grade}")
```
